## Sentiment Analysis

The sentiment property returns a namedtuple of the form Sentiment(polarity, subjectivity). The polarity score is a float within the range [-1.0, 1.0]. 

**The subjectivity is a float within the range [0.0, 1.0] where 0.0 is very objective and 1.0 is very subjective.**

```python
testimonial = TextBlob ("Textblob is amazingly simple to use. What great fun!")
testimonial.sentiment
testimonial.sentiment.polarity
```

## Words Inflection and Lemmatization

Each word in TextBlob.words or Sentence.words is a Word object with useful methods, such as for **word infelction**

```python
sentence = TextBlob("Use 4 space bar per indentation level.")
sentence.words
sentence.words[2].singularize()
sentence.words[-1].pluralize()
```

Words can be lemmatized by calling the `lemmatize` method

```python
from textblob import Word
w = Word ("octopi")
w.lemmatize()
w = Word("went")
w.lemmatize("v") # verb
 
```

## WordNet Integration

I can access the synsets for a Word via the synsets property or the `get_synsets` mwthod.

```python
from textblob import Word
from textblob.wordnet import VERB
word = Word ("octopus")
word.synsets
Word("hack").get_synsets(pos=VERB)
```

I can access the definitions for each synset via the definition prepeerty or the `define()` method.

```python
Word ("Octopus").definitions
```

I can also create synsets directly

```python
from textblob.wordnet import Synset
octopus = Synset ("octopus.n.02")
shrimp = Synset("shrimp.n.03")
octopus.path_similarity(shrimp)
```

## Get Word and Noun Phrase Frequencies

The first way to get the frequency of a word or noun phrase in a TextBlob is through the `word_counts` dictionary.

```python
monty = TextBlob("We are no longer the Knights who say Ni. "
                    "We are now the Knights who say Ekki ekki ekki PTANG.")
monty.word_counts['ekki']
```

This way, the search will *not* be case sensitive, and words that are not found will have a frequency of 0. 

(Case-sensitive means Python treats uppercase and lowercase letters as different.)

The second way is to use the `count()` method.

```python
monty.words.count("ekki")
```

```python
monty.words.count('ekki', case_sensitive=True)
```

Each of these methods can also be used with noun phrases.


```python
wiki.noun_phrases.count("python")
```

## TextBlobs Are Like Python Strings!

I can use Python's substring syntax.

```python
zen[0:19]
```

I can use common string methods.

```python
zen.upper()
zen.find("Simple")
```

I can make comparisons between TextBlobs and strings.

```python
apple_blob = TextBlob("apples")
banana_blob = TextBlob("bananas")
apple_blob < banana_blob
apple_blob == "apples"
```

I can concatenate and interpolate TextBlobs and strings.

```python
apple_blob + " and " + banana_blob
"{0} and {1}".format(apple_blob, banana_blob)
```

## n-grams

The `TextBlob.ngrams() ` method returns a list of tuples of **n** successive words.

```python
blob = TextBlob ("Now is better than never.")
blob.ngrams(n=3)
```
